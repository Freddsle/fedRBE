# Inspect the feature overlap

Run from `evaluation_data/ovarian_cancer/` after the collapsed per-study matrices
are available. The selected studies are the live `before/GSE*/` folders.


```python
import pandas as pd
import sys
import matplotlib.pyplot as plt
from pathlib import Path

sys.path.insert(0, str(Path('../../evaluation_utils').resolve()))
import upset_plot

# suppress warnings
import warnings
warnings.filterwarnings('ignore')
```

## GenBank accession IDs


```python
id_lists = {}

for matrix in sorted(Path('before').glob('GSE*/intermediate/expr_GBacc_MaxVar.tsv')):
    study = matrix.parents[1].name
    features = pd.read_csv(matrix, sep='\t', usecols=['row_ID'])
    id_lists[f'{study}_1'] = features.rename(columns={'row_ID': 'GB_ACC'})

if not id_lists:
    raise FileNotFoundError('No collapsed ovarian study matrices found under before/GSE*/.')
```

```python
id_lists['GSE69428_1'].head()
```

```python
upset_plot.generate_upset_plot(
    id_lists,
    colname='GB_ACC',
    title='Upset plot of ovarian cancer ovarian_cancer data, GB ids',
    categories=list(id_lists.keys()),
    save_plot=True,
    path_svg='preprocessing_info/upset_plot_GBACC.svg',
)
```
